```python
import re
from gensim.models import Word2Vec
```

```python
text = open("document3.txt", "r").read()
```

```python
text = text.lower()
```

```python
text = re.sub(r"[^a-z\s]", "", text)
```

```python
words = text.split()
```

```python
sentences = [words]
```

```python
model = Word2Vec(
    sentences, 
    vector_size = 50,
    window = 2,
    min_count = 1,
    sg = 0,
    epochs = 10
)
```

```python
word = "influenza"
print("Word: ", word)
print("Vector: ", model.wv[word])
print("Similar words: ", model.wv.most_similar(word, topn=5))
```
